```python
import numpy as np

def wiener(T=1.0, dT = 0.0001):
  N = int(1 / dT)
  t = np.linspace(0, T, N, dtype=float)
  dW = np.sqrt(dT) * np.random.randn(N)
  W = np.cumsum(dW, dtype=float)
  return t, W
```

```python
import matplotlib.pyplot as plt

N_TRAJECTORIES = 1000

np.random.seed(42)

plt.figure(figsize=(12, 6))

colors = ['red', 'blue', 'green', 'magenta', 'yellow', 'cyan']

for i in range(N_TRAJECTORIES):
    t, W = wiener()
    plt.plot(t, W, linewidth=0.2, color=colors[i % len(colors)])

plt.grid(which='major', linestyle='-')
plt.grid(which='minor', linestyle='--')
plt.minorticks_on()

plt.title('1000 trajectories of Wiener Process')
plt.xlim(0, 1)
plt.xlabel('$t$')
plt.ylabel('$W(t)$')
plt.savefig("wiener.png")
```
